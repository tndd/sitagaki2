```python
import pandas_ta as ta
import polars as pl

# Date列をインデックスに設定
pandas_df = ohlcv.df.to_pandas().set_index("Date")
print(pandas_df)

# カスタム戦略を作成
my_strategy = ta.Strategy(
    name="base",
    ta=[
        # trend
        {"kind": "sma", "length": 5},
        {"kind": "sma", "length": 25},
        {"kind": "sma", "length": 75},
        # momentum
        {"kind": "rsi"},
        {"kind": "roc"},
        # volatility
        {"kind": "atr"},
        {"kind": "bbands"},
        # volume
        {"kind": "obv"},
        {"kind": "vwap"},
    ],
)

# 戦略を適用
pandas_df.ta.strategy(my_strategy)
# print(pandas_df[75:])

ta_pldf = pl.from_pandas(pandas_df[75:].reset_index())
ta_pldf
```
